## 1. Project Context

This notebook is the first step of the **Olist Revenue Intelligence project**.

The objective of this notebook is to understand the **structure of the Olist e-commerce dataset** before performing any analytical work.

More specifically, this phase focuses on:

- understanding the relational schema of the dataset
- identifying primary keys and relationships between tables
- determining the analytical grains of the data
- locating where revenue is actually stored in the dataset

This step is essential to ensure that subsequent analyses and business metrics are built on a **correct understanding of the underlying data model**.

## 2. Data Loading

In this section, we load the raw Olist dataset tables into Python using pandas.

The goal is to import all available tables in order to explore their structure and understand the relationships between them.

The dataset contains nine CSV files describing different aspects of the e-commerce platform, including customers, orders, products, payments, reviews, and sellers.

```python
import pandas as pd
import numpy as np
```

```python
data_path = "../data/raw/"
```

```python
customers = pd.read_csv(data_path + "olist_customers_dataset.csv")
geolocation = pd.read_csv(data_path + "olist_geolocation_dataset.csv")
order_items = pd.read_csv(data_path + "olist_order_items_dataset.csv")
payments = pd.read_csv(data_path + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(data_path + "olist_order_reviews_dataset.csv")
orders = pd.read_csv(data_path + "olist_orders_dataset.csv")
products = pd.read_csv(data_path + "olist_products_dataset.csv")
sellers = pd.read_csv(data_path + "olist_sellers_dataset.csv")
translation = pd.read_csv(data_path + "product_category_name_translation.csv")
```

```python
tables = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "translation": translation
}
```

## 3. Schema Inspection

Before performing any analytical work, it is necessary to understand the structure of the dataset.

In this section, we inspect each table to identify:

- number of rows and columns
- available variables
- data types
- potential missing values

This step helps us understand the overall structure of the Olist dataset and identify which tables contain the most relevant information for the analysis.

```python
for name, df in tables.items():
    
    print("="*60)
    print(f"TABLE: {name}")
    print("="*60)
    
    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())
    
    print("\nData types:")
    print(df.dtypes)
    
    print("\nMissing values (%):")
    print((df.isna().mean()*100).round(2).sort_values(ascending=False))
    
    print("\nPreview:")
    display(df.head())
```

## 4. Key Validation

In this section, we verify whether the columns that appear to be identifiers are actually unique.

This step is essential to:
- identify valid primary keys
- avoid incorrect joins
- understand which tables behave as dimensions and which ones behave as transactional tables

In a relational dataset, assuming uniqueness without checking it can lead to duplicated rows and incorrect business metrics.

```python
key_checks = {
    "customers": ["customer_id", "customer_unique_id"],
    "orders": ["order_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "translation": ["product_category_name"]
}

for table_name, keys in key_checks.items():
    df = tables[table_name]
    
    print("=" * 60)
    print(f"TABLE: {table_name}")
    print("=" * 60)
    
    for key in keys:
        n_rows = len(df)
        n_unique = df[key].nunique(dropna=False)
        is_unique = n_rows == n_unique
        
        print(f"{key}:")
        print(f"  rows      = {n_rows}")
        print(f"  unique    = {n_unique}")
        print(f"  is_unique = {is_unique}")
        print()
```

## 5. Relationship and Cardinality Analysis

After validating the primary keys, the next step is to understand the relationships between the tables.

In a relational dataset, different tables may be connected through **one-to-one** or **one-to-many relationships**.

Understanding these cardinalities is essential to avoid incorrect joins that could artificially duplicate rows and distort business metrics.

In this section we analyze how many:

- orders each customer has
- items each order contains
- payments are associated with each order
- reviews exist per order

```python
orders_per_customer = orders.groupby("customer_id")["order_id"].nunique()

print("Orders per customer:")
display(orders_per_customer.describe())
```

```python
items_per_order = order_items.groupby("order_id").size()

print("Items per order:")
display(items_per_order.describe())
```

```python
payments_per_order = payments.groupby("order_id").size()

print("Payments per order:")
display(payments_per_order.describe())
```

```python
reviews_per_order = reviews.groupby("order_id").size()

print("Reviews per order:")
display(reviews_per_order.describe())
```

```python
print("Unique orders in reviews:", reviews["order_id"].nunique())
print("Total review rows:", len(reviews))
print("Orders with more than one review:", (reviews_per_order > 1).sum())
```

### Methodological note on reviews

Reviews are linked to `order_id`, which means they belong naturally to the **order level**, not to the item level.

The cardinality check also shows that while most orders have a single review entry, a small number of orders have multiple review rows.

As a result, any downstream integration of customer feedback should be done through an **order-level aggregated review table**, rather than by merging the raw reviews table directly into an item-level fact table.

## 6. Order Lifecycle Exploration

Before building revenue metrics, it is necessary to understand the lifecycle of an order.

Orders can have different statuses such as delivered, shipped, canceled, unavailable, or processing.

Not all of these statuses should be treated as completed business transactions.

In this section, we examine the distribution of order statuses in order to determine which subset of orders should be retained for the revenue analysis.

```python
orders["order_status"].value_counts(dropna=False)
```

```python
orders["order_status"].value_counts(normalize=True, dropna=False).round(4)
```

```python
order_status_summary = (
    orders["order_status"]
    .value_counts(dropna=False)
    .reset_index()
)

order_status_summary.columns = ["order_status", "n_orders"]
order_status_summary["share"] = (
    order_status_summary["n_orders"] / order_status_summary["n_orders"].sum()
).round(4)

order_status_summary
```

```python
orders_delivered = orders[orders["order_status"] == "delivered"].copy()
```

```python
print("Total orders:", len(orders))
print("Delivered orders:", len(orders_delivered))
print("Share delivered:", round(len(orders_delivered) / len(orders), 4))
```

## 7. Revenue Source Identification

At this stage, the objective is to identify where revenue is actually stored in the Olist dataset.

There are two main monetary sources in the data:

- `price` and `freight_value` in the `order_items` table
- `payment_value` in the `payments` table

These variables do not belong to the same analytical grain, so they must be analyzed carefully.

The purpose of this section is to:
- identify the relevant monetary columns
- compute their aggregate values
- compare item-level revenue with payment-level revenue
- determine the most appropriate revenue definition for the project

```python
monetary_columns = {
    "order_items": ["price", "freight_value"],
    "payments": ["payment_value"]
}

for table_name, cols in monetary_columns.items():
    print("=" * 60)
    print(f"TABLE: {table_name}")
    print("=" * 60)
    print(tables[table_name][cols].describe())
    print()
```

```python
delivered_order_ids = set(orders_delivered["order_id"])
```

```python
order_items_delivered = order_items[
    order_items["order_id"].isin(delivered_order_ids)
].copy()

order_items_delivered["item_revenue"] = order_items_delivered["price"]
order_items_delivered["total_revenue"] = (
    order_items_delivered["price"] + order_items_delivered["freight_value"]
)

print("Delivered order_items revenue summary")
print("Product revenue:", round(order_items_delivered["item_revenue"].sum(), 2))
print("Total revenue (price + freight):", round(order_items_delivered["total_revenue"].sum(), 2))
```

```python
payments_delivered = payments[
    payments["order_id"].isin(delivered_order_ids)
].copy()

print("Delivered payments summary")
print("Payment collected:", round(payments_delivered["payment_value"].sum(), 2))
```

```python
product_revenue = order_items_delivered["item_revenue"].sum()
total_revenue = order_items_delivered["total_revenue"].sum()
payment_collected = payments_delivered["payment_value"].sum()

revenue_comparison = pd.DataFrame({
    "metric": [
        "Product revenue",
        "Total revenue (price + freight)",
        "Payment collected"
    ],
    "value": [
        round(product_revenue, 2),
        round(total_revenue, 2),
        round(payment_collected, 2)
    ]
})

revenue_comparison
```

```python
difference = total_revenue - payment_collected
relative_difference = (difference / payment_collected) * 100

print("Difference:", round(difference, 2))
print("Relative difference (%):", round(relative_difference, 4))
```

### Revenue Definition Decision

The comparison between item-side revenue and payment-side revenue shows that:

- `SUM(price + freight_value)` on delivered orders is very close to `SUM(payment_value)`
- the relative difference is negligible

Therefore, the main revenue definition retained for the project is:

**Revenue = SUM(price + freight_value)**

computed on **delivered orders**, with **order_items** as the main analytical grain.

## 8. Preliminary Business Variables

Before building the analytical dataset, it is useful to create a few core business variables that will support future analyses.

These variables help transform raw transactional fields into more interpretable analytical features.

In this section, we create variables related to:

- revenue
- order timing
- delivery timing
- delivery performance

```python
date_cols = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    orders_delivered[col] = pd.to_datetime(orders_delivered[col], errors="coerce")
```

```python
order_items_delivered["item_revenue"] = order_items_delivered["price"]
order_items_delivered["total_revenue"] = (
    order_items_delivered["price"] + order_items_delivered["freight_value"]
)
```

```python
orders_delivered["order_month"] = (
    orders_delivered["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
orders_delivered["delivery_lead_time_days"] = (
    orders_delivered["order_delivered_customer_date"]
    - orders_delivered["order_purchase_timestamp"]
).dt.days
```

```python
orders_delivered["estimated_vs_actual_days"] = (
    orders_delivered["order_delivered_customer_date"]
    - orders_delivered["order_estimated_delivery_date"]
).dt.days
```

```python
orders_delivered["is_late"] = (
    orders_delivered["estimated_vs_actual_days"] > 0
).astype(int)
```

### Methodological note on delivery variables

The variables `delivery_lead_time_days`, `estimated_vs_actual_days`, and `is_late` are naturally defined at the **order level**.

They are useful for delivery-risk analysis and later predictive modeling, but they should not be interpreted as pure item-level metrics when the analytical grain shifts to `order_items`.

This distinction will matter in later notebooks when separating:
- item-level revenue structure analysis
- order-level operational execution analysis

```python
orders_delivered[
    [
        "order_id",
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "order_month",
        "delivery_lead_time_days",
        "estimated_vs_actual_days",
        "is_late"
    ]
].head()
```

```python
print("Delivery delay summary:")
display(orders_delivered["delivery_delay_days"].describe())

print("\nEstimated vs actual summary:")
display(orders_delivered["estimated_vs_actual_days"].describe())

print("\nLate delivery rate:")
print(round(orders_delivered["is_late"].mean() * 100, 2), "%")
```

## 9. Analytical Grain Decision

At this stage, the main methodological choices of the project can be formalized.

Based on the previous exploration, the following decisions are retained:

- the main analytical population will be restricted to **delivered orders**
- the main revenue definition will be **price + freight_value**
- the project will rely on **two complementary analytical layers**

### Item-level analytical layer
A first analytical layer will be built at the `order_items` grain in order to support:
- revenue structure analysis
- category-level analysis
- seller-level analysis
- portfolio analysis

### Order-level analytical layer
A second analytical layer will be built at the `orders` grain in order to support:
- delivery performance indicators
- lateness analysis
- customer feedback analysis through reviews
- future predictive modeling

This distinction is necessary because revenue is naturally decomposed at item level, whereas delivery outcomes and review signals are naturally defined at order level.

### 9.1 Decision Summary

The analytical framework retained for the project is the following:

- **Population:** delivered orders only
- **Revenue definition:** `SUM(price + freight_value)`
- **Main customer identifier:** `customer_unique_id`

Two analytical grains will be used downstream:

- **Item-level fact table:** one row per order item, for revenue and portfolio analysis
- **Order-level fact table:** one row per order, for delivery KPIs, review analysis, and predictive modeling

The main transactional tables are therefore:
- `order_items` for detailed commercial structure
- `orders` as the main order-level reference table

```python
analytical_decisions = pd.DataFrame({
    "decision_area": [
        "Population",
        "Revenue definition",
        "Customer identifier",
        "Item-level analytical layer",
        "Order-level analytical layer",
        "Main transactional table",
        "Order-level reference table"
    ],
    "retained_choice": [
        "Delivered orders only",
        "SUM(price + freight_value)",
        "customer_unique_id",
        "One row per order item (revenue, category, seller, portfolio analysis)",
        "One row per order (delivery KPIs, reviews, predictive modeling)",
        "order_items",
        "orders"
    ]
})

analytical_decisions
```
